# Conda + NumPy 环境测试

运行下方单元格，验证 Python、NumPy 以及 Jupyter Notebook 是否正常。

```python
import sys
import numpy as np

a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
result = a @ b

print(f"Python: {sys.version.split()[0]}")
print(f"NumPy: {np.__version__}")
print("Matrix product:")
print(result)
assert np.array_equal(result, np.array([[19, 22], [43, 50]]))
print("TEST PASSED")
```
