# 括号生成

```python
# n = int(input())
n = 3
ans = []


'''
思路：dfs回溯
除非到达最后一个字符
右括号数目 < 左括号
) (

先放左括号，再放右括号

终止条件：len(str) == 2*n
'''

def dfs(path, left, right): 
    if len(path) == 2*n:
        ans.append(path)
        return
    
    if left < n:
        dfs(path+"(", left+1, right)

    if right < left:
        dfs(path+")", left, right+1)

dfs("", 0, 0)
print(ans)
```
