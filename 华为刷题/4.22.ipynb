# 第一题：统计二叉树中平衡路径的数量

判断相等，is 更符合 python 规范

任意区间和为0，即：prefix[j] - prefix[i] == 0。就是我们要的平衡路径。

平衡路径定义：
* 长度>=2
* 任意节点开始，一段区间

思路：统计这段路程的prefix_count[number]。到底出现了几次，就加几次

```python
from collections import deque, defaultdict

# 二叉树定义
class TreeNode:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

# 输入层序遍历的二叉树
# 用队列建树。先进先出。
def build_tree(arr):
    # 空列表 或者 root为null
    if not arr or arr[0] is None:
        return None
    
    root = TreeNode(arr[0])

    # 利用队列build tree
    q = deque([root]) # deque需要可迭代对象（list）

    i = 1
    while q and i < len(arr):
        node = q.popleft()

        # 左子树
        if  arr[i] is not None:
            node.left = TreeNode(arr[i])
            q.append(node.left)
        
        i += 1

        # 右子树
        if  i<len(arr) and arr[i] is not None:
            node.right = TreeNode(arr[i])
            q.append(node.right)
        
        i += 1

    return root

def count_balanced_paths(arr):
    root = build_tree(arr)
    ans = 0

    prefix_count = defaultdict(int)
    prefix_count[0] = 1

    # 回溯。通过 dfs遍历树的路径，过程中记录前缀和。

    def dfs(node, cur_sum):
        # 使用外层变量
        nonlocal ans 

        if not node:
            return
        
        cur_sum += node.val

        ans += prefix_count[cur_sum]

        # 当node为0，本身和为0，会被认为是一条“平衡路径”。这里去除掉
        if node.val == 0:
            ans -= 1

        # 因为只能单向 走左/走右，左右子树互不干扰
        prefix_count[cur_sum] += 1
        dfs(node.left, cur_sum)
        dfs(node.right, cur_sum)
        prefix_count[cur_sum] -= 1 


    dfs(root, 0)

    return ans

# === io ===

import ast

arr = ast.literal_eval(input())
print(count_balanced_paths(arr))
```
